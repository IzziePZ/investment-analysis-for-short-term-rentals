# Introduction

Airbnb has become one of the largest short-term rental platforms in the world, providing hosts with opportunities to generate income while offering travelers a wide range of accommodation options. Because listing prices are influenced by numerous factors—including location, property characteristics, amenities, reviews, and availability—determining an appropriate nightly rate can be challenging.

The goal of this analysis is to explore Airbnb listing data and develop machine learning models capable of predicting nightly prices. Through exploratory data analysis (EDA), we will examine the relationships between listing features and pricing, identify key drivers of price variation, and prepare the data for predictive modeling.

To estimate nightly prices, two regression models will be evaluated: Support Vector Regression (SVR) and Gradient Boosting Regressor. Model performance will be assessed using standard regression metrics, including Root Mean Squared Error (RMSE), Mean Absolute Error (MAE), and the coefficient of determination (R²).

The insights gained from this analysis will serve as the foundation for a broader investment-focused study comparing Airbnb markets across major U.S. cities and evaluating potential revenue opportunities for short-term rental investors.


Dataset was downloaded from kaggle. Link: https://www.kaggle.com/datasets/youssefkhaled117/airbnb
The dataset contains information about listing characteristics, host details, geographic location, pricing structure, availability, and customer feedback.

## Predictive Pricing Model

Since the target variable, nightly price, is numerical, this problem will be approached using regression techniques. Two regression models will be developed and evaluated:

1. **Support Vector Regression (SVR)** – SVR is capable of capturing complex, non-linear relationships between listing characteristics and pricing, making it well-suited for Airbnb datasets.

2. **Gradient Boosting Regressor** – Gradient Boosting is an ensemble learning method that builds a series of decision trees to improve prediction accuracy. It is particularly effective at modeling non-linear relationships and interactions among features, often achieving strong performance on structured tabular data.

The performance of both models will be evaluated and compared using standard regression metrics such as Root Mean Squared Error (RMSE), Mean Absolute Error (MAE), and the coefficient of determination (R²). The best-performing model will be used to generate pricing recommendations based on listing attributes and market conditions.
